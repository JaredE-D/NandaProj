V6 redirects the remaining hours instead of producing an uninterpretable null at
hour 19.

Everything here runs at **identical layer, `k`, and probe methodology** as the
main measurement, or the comparison means nothing.

- **Positive control** (sentiment / topic) -- known verbalizable. *This is what
  licenses any W1 claim.*
- **Negative control** (token position, sequence length) -- sets the floor.
- **Covariance-matched random directions.** Activation space is strongly
  anisotropic; an isotropic null inflates significance and a reviewer will say so.
- **Confound check on `v_acc`:** within-topic, and MMLU -> TriviaQA transfer.

**V5:** the null distribution for rho **and** the effect-size threshold are fixed
*before* looking at `rho(v_acc)`. Both, at the same moment (R10).

**V6:** positive control separates from negative control. If it does not, the lens
cannot discriminate at this layer/`k`, **no null is interpretable -- stop.**
