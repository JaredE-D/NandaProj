Defined against PLAN.md 2.1. The existing pipeline builds an **external** bridge
from `v_acc` to `v_verb` (probe -> isotonic -> inverse transfer -> alpha). This
asks whether an **internal** one exists:

> Inject along `v_acc` into J-space. Does verbalized confidence move -- *without*
> the probe/isotonic/alpha scaffold?

- **Moves** -> the wiring is present and unused. W1 collapses toward W2; the
  external scaffold substitutes for a path the model already has.
- **Doesn't move** -> causal evidence for genuine absence rather than an
  instrument null.

**V9:** injection produces a **monotone dose-response** in verbalized confidence.
One dose is not a result; the sweep is the result.
