```python
def rho(v, basis):
    """Fraction of v's squared norm inside the span of `basis` (orthonormal rows)."""
    import numpy as np
    v = np.asarray(v, dtype=np.float64)
    proj = basis.T @ (basis @ v)
    return float(proj @ proj) / float(v @ v)
```
