# 04e — Reconcile: the probe turns at L19–21, the lens drops at L25

04d and 04 disagree on *where* the deceptive prompt overwrites the answer. The H-fitted
mean-difference truth probe, read on D residuals of lying items, says the truth through
L16 and the lie from L19–21 on. The J-lens curves say `P_J(a_H)` under D stays near 1 on
Yes-true items through L24 and drops at L25 -- and near 0 on No-true items at every layer,
because the lens decodes `" Yes"` from L19 and `" No"` only from L25 in every condition.
Two instruments, two layers, one residual stream.

**The measurement, stated once:**

> Section 1 puts both readings on the same axis: the signed projection of every D
> residual onto the H-fitted truth axis, per layer, split by polarity and by whether the
> model lied, beside the lens's `P_J(a_H)` on the same items. Section 2 asks the PLAN2
> §4.5 question at every fitted layer: what fraction of each probe direction lies inside
> the top-k right singular subspace of `J_ℓ`, against a covariance-matched null.

| | |
|---|---|
| section 1 | CPU. Reloads `slot_residuals_*.npz`, `slot_emitted_*.json` (04d) and `belief_readout_all_*.npz` (04) from `results/`. No model |
| section 2 | GPU. Needs the model and the fitted lens for `J_ℓ`, plus one residual pool per layer for the covariance null (PLAN2 §12.4) |
| helpers | [`probe.signed_projection`, `probe.subspace_mass`](../src/nandaproj/probe.py), numpy, tested locally |

## Language

Nothing here is a belief. The most a number below licenses is "the true answer is
linearly decodable from `h_ℓ` under D" and "that direction has fraction ρ of its mass
inside J-space(k)". The lens reading is "the honest-answer token carries mass in the
J-lens readout at layer ℓ", nothing more.
